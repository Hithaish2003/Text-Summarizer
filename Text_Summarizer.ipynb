```python
from nltk.tokenize import word_tokenize
import re
from spacy.lang.en.stop_words import STOP_WORDS
from nltk.stem import PorterStemmer
import spacy
from transformers import pipeline
import pickle as pk
def nlp_pipeline(txt):
    text = re.sub(r"\s+", " ", txt)
    text = re.sub(r"https?://\S+|www\.\S+", "", text)
    text = re.sub(r"[^a-zA-Z0-9.,!?\s]", "", text)
    nlp=spacy.load('en_core_web_sm')
    doc=nlp(text)
    data=[i for i in doc if i not in STOP_WORDS and i.text.isalpha()]
    print(data)
    stems=[]
    lemmas=[]
    pos=[]
    ner=[]
    ps=PorterStemmer()
    for i in data:
        stems.append(ps.stem(i.text))
        lemmas.append(i.lemma_)
        pos.append((i,i.pos_))
    for i in doc.ents:
        if i.label_:
            ner.append((i,i.label_))
    summarizer = pipeline("summarization")
    with open('model.pkl', 'wb') as f:
        pk.dump(summarizer, f)
    summary = summarizer(txt, max_length=100, min_length=30, do_sample=False)
    print("Root words using the Stemming:\n",stems)
    print("Root words using the Lemmatization:\n",lemmas)
    print("POS Tags:\n",pos)
    print("Named Entity Recognization:\n",ner)
    print(summary[0]['summary_text'])

t="""Text summarization is a key NLP task that condenses long documents into shorter versions while preserving meaning.
It can be done using extractive methods, which select key sentences, or abstractive methods, which generate new sentences.
Recent transformer-based models like BART and PEGASUS have significantly improved abstractive summarization quality.
This pipeline demonstrates both methods using Python libraries."""
nlp_pipeline(t)
```
